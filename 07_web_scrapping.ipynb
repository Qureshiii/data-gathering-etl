```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
import asyncio, sys, warnings

if sys.platform == "win32":
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", DeprecationWarning)
        asyncio.set_event_loop_policy(asyncio.WindowsProactorEventLoopPolicy())
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}
webpage = requests.get('https://www.ambitionbox.com/list-of-companies?page=1',headers=headers).text
```

            Extracting company Names

```python
soup = BeautifulSoup(webpage,'lxml')
```

```python
company = soup.find_all('div',class_='company-content-wrapper')
```

```python
name = []
rating = []
review = []
ctype = []
hq = []
old = []
employees = []

for i in company:

    name.append(i.find('h2').text.strip())
    rating.append(i.find('p',class_='rating').text.strip())
    review.append(i.find('a',class_='review-count').text.strip())
    ctype.append(i.find_all('p',class_='infoEntity')[0].text.strip())
    hq.append(i.find_all('p',class_='infoEntity')[1].text.strip())
    old.append(i.find_all('p',class_='infoEntity')[2].text.strip())
    employees.append(i.find_all('p',class_='infoEntity')[].text.strip())
    

d={'name':name,'rating':rating,'review':review,'type':ctype,'head_quarter':hq,'old':old,'employees':employees}

df = pd.DataFrame(d)
df
```
